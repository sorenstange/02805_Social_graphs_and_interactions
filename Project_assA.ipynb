### Central Idea
We would like to investigate and compare the two worlds as created in "Lord of the rings" and "Harry Potter" and related works. The two creations are hugely popular worldwide and some similarities between the stories are apparent (e.g. the fight against evil is a main theme in both, and some of the figures are comparable e.g. Dumbledore and Gandalf), and we wish to explore what similarities and differences can be found between characters, relations and structure in a network. 

We create two networks - one for Lord of the rings and one for harry potter - with character data downloaded from the respective wiki fandoms. Here nodes are characters and edges are links between character sites, while the content on the pages is downloaded as well. Different characteristica (relations, Hogwarts house, race etc.) are also marked.

We use the graphs to explore role equivalence (is Gandalf comparable to Dumbledore, Sauron to Voldemort, Frodo to Harry Potter), and compare role similarities to identify figures that occupy simular positions is their respective networks.
We compare the social structure of the networks, to see if one universe is more centrallized, hierachical or egalitarian in terms of interactions. 
We also wish to do a sentiment analysis on the content of the character pages to explore if we can see how negativity propagates (e.g. from Voldemort and Sauron), and compare polarization of the two universes.
