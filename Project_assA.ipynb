### Central Idea
We would like to investigate and compare the two worlds as created in the universe of "Lord of the rings" and "Harry Potter". The two universes are hugely popular worldwide and some similarities between the stories are apparent (e.g. the fight against evil is a main theme in both, and some of the figures are comparable e.g. Dumbledore and Gandalf), and we wish to explore what similarities and differences can be found between characters, relations and structure of the universe when creating a network of characters and relations. 

We create two networks - one for Lord of the rings and one for Harry Potter - with character data downloaded from the respective wiki fandoms. Here nodes are characters and edges are links between character sites, while the content on the pages is downloaded as well. Different characteristica ( Hogwarts house, race) are also downloaded.

We use the graphs to explore role equivalence (is Gandalf comparable to Dumbledore, Sauron to Voldemort, Frodo to Harry Potter), and compare role similarities to identify figures that occupy simular positions in their respective networks.
We compare the social structure of the networks, to see if one universe is more centrallized, hierachical or egalitarian in terms of interactions. 
We also wish to do a sentiment analysis on the content of the character pages to explore if we can determine how negativity propagates (e.g. from Voldemort and Sauron), and compare polarization of the two universes.

Our goal is to explore how these two worlds are structured through their characters and relationships and how similar they are.

***Stats fra data***

We create two character networks, one for each universe, using data downloaded from the official fandom wikis.
In these networks, each node represents a character, and edges are formed between characters whose pages link to each other.
Along with the network data, we also collect text and character attributes — such as race for The Lord of the Rings or Hogwarts house for Harry Potter.
This gives us both relational and textual information to analyze.

***Prelim grafer***
